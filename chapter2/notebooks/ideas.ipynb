## What has changed in fiber optics recently

| topic | why it matters | where it connects in Chapter 2 |
|---|---|---|
| **Coherent detection + DSP** | Chromatic dispersion and PMD are now undone digitally, so DCF disappears from new links | dispersion compensation, power fading |
| **Hollow-core anti-resonant fibers** | < 0.1 dB/km reported, 1/3 lower latency, ~1000× lower nonlinearity | special fibers, Rayleigh limit |
| **Space-division multiplexing** | multicore and few-mode fibers multiply capacity again (petabit/s per fiber in the lab) | LP modes, coupled-mode theory |
| **Multi-band (O to U)** | low-water-peak fibers + new amplifiers use > 50 nm beyond the C-band | attenuation spectrum, OH peak |
| **Distributed acoustic sensing on telecom cables** | φ-OTDR turns submarine and urban cables into seismometers and traffic sensors | OTDR, Rayleigh backscatter |
| **PAM-4 / IM-DD in data centres** | 100–200 Gb/s per lane over short SMF, close to λ₀ = 1310 nm | B²L limit, power fading |
| **Quantum key distribution** | single photons share fibers with classical channels, so Raman noise and FWM matter | nonlinear effects, Raman |

### A taste of the nonlinear Shannon limit

Amplified links add ASE noise ($\propto$ number of spans), and nonlinear interference grows as $P^3$ (the Gaussian-noise
model). The SNR, $\mathrm{SNR}=P/(P_\text{ASE}+\eta P^3)$, therefore has an **optimum launch power**, and the capacity
per fiber has a ceiling.

```python
#| label: fig-gn
#| fig-cap: "Toy GN-model: SNR versus launch power per channel for links of 10, 20 and 40 spans of 80 km. Below the optimum the link is noise-limited, above it nonlinearity-limited."
P = np.logspace(-4, -1.5, 300)          # W per channel
h_nu, NF, G = fl.h * 193.4e12, 10 ** 0.5, 10 ** (80 * 0.2 / 10)
B, eta1 = 32e9, 1.0e3                    # symbol bandwidth, NLI coefficient per span [1/W²]
fig, ax = plt.subplots()
for k, Ns in enumerate([10, 20, 40]):
    ase = Ns * NF * h_nu * G * B
    snr = P / (ase + Ns * eta1 * P**3)
    ax.plot(10 * np.log10(P / 1e-3), 10 * np.log10(snr), color=fl.PALETTE[k], label=f"{Ns} spans ({Ns*80} km)")
    i = snr.argmax(); ax.plot(10*np.log10(P[i]/1e-3), 10*np.log10(snr[i]), "o", color=fl.PALETTE[k])
ax.set_xlabel("launch power per channel (dBm)"); ax.set_ylabel("SNR (dB)"); ax.legend()
plt.show()
```

## Improvements to the course

1. **Live figures instead of screenshots.** Many slides show plots taken from books (b–V curves, $D(\lambda)$, loss
   spectrum, eye diagrams). Every one of them is now regenerated in this book by a few lines of code. Students can change a
   parameter and see the consequence, which builds intuition faster than a static picture.
2. **"Predict, then compute."** For each live panel, ask the class to predict the outcome before moving the slider
   (e.g. *"What happens to the eye if D changes sign?"*). This is quick, engaging and well suited to peer instruction.
3. **Exercises with automatic checking.** The worked examples are all reproduced numerically, so they can become
   exercises with randomised numbers (e.g. with [quarto-live](https://r-wasm.github.io/quarto-live/) or Pyodide, Python runs
   in the browser and the answer can be checked on the spot). The problem set `Chapter2_Questions_Problems.pdf` is a natural
   source.
4. **One running example through the chapter.** Use the same standard SMF (a = 4.1 µm, 3.5 % GeO₂) from modes → MFD →
   dispersion → bend loss → link budget → eye diagram. Students then see how the design choices are coupled (e.g. a
   smaller core shifts λ₀ but raises bend loss).
5. **Connect with the online link simulator** already used in the course
   (see `opticallinksimulator.txt`) and with the in-browser simulator of @sec-bitrate.
6. **Shorten the sensor part, deepen the "why".** The gyroscope history slides (301–309) could become a single
   timeline, freeing time for the Sagnac calculation and the OTDR trade-offs, which are more transferable skills.
